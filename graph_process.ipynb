```python
import torch
from torch_geometric.data import Data, Dataset
```

```python
def get_mapping_dict(mapping):
  rec = {}

  for i in mapping.split(";"):
    key = i.split(":")[0].strip()
    value = i.split(":")[1].strip()

    start = int(value.split("-")[0])
    end = int(value.split("-")[1])

    rec[key] = [start, end]
  
  return rec

def create_graph_object(graph_str, target_node, var_mapping, embedding, label):
  print("graph_str")
  print(graph_str)
  penman_graph = penman.decode(graph_str)

  instances = penman_graph.instances()
  edges = penman_graph.edges()

  node2index = {}
  index2node = {}

  mapping_rec = get_mapping_dict(var_mapping)

  node_data = torch.tensor([])
  curr_index = 0

  target_index = -1
  edges_rec = []
  for i in range(len(instances)):
    print("instance", instances[i])
    s = instances[i].source

    start, end = mapping_rec.get(s, [0,0])

    if start != 0:
      start -= 1
      k = embedding[start:end]
      node_data = torch.cat((node_data, torch.sum(k, dim=0).unsqueeze(0)))
    else:
      node_data = torch.cat((node_data, torch.sum(embedding, dim=0).unsqueeze(0)))

    node2index[s] = curr_index
    curr_index += 1

    if s == target_node:
      target_index = curr_index - 1
  
  for e in edges:
    print(e)
    source = e.source
    target = e.target
    edges_rec.append([node2index[source], node2index[target]])
    edges_rec.append([node2index[target], node2index[source]])

  # edges_rec.append([target_index, node2index['aspect']])
  edges_data = torch.tensor(edges_rec).t().contiguous()
  print("node2index", node2index)
  return Data(x=node_data, edge_index=edges_data, y=torch.tensor(label), target_node=torch.tensor(node2index[target_node]))

```
